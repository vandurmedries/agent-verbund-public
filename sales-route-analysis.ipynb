# Sales route prioritization
Analyst-coded ordinal scores. No measured conversion or revenue. See SALES-ROUTE-STUDY.md for sources and gates.

```python
import json
from pathlib import Path
data=json.loads(Path('sales-route-scores.json').read_text())
for row in data['rows']:
    score=sum(s*w for s,w in zip(row['scores'],data['weights'].values()))/5
    assert score==row['priority_score']
    print(row['id'],score)
# Alternative priorities: demand-heavy and delivery-heavy
for weights in ([40,15,20,10,15],[15,40,20,10,15]):
    ranked=sorted([(sum(s*w for s,w in zip(r['scores'],weights))/5,r['id']) for r in data['rows']],reverse=True)
    print(weights,ranked)
```
